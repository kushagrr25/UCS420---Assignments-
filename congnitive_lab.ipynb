### Assignment 5.2: Add numbers divisible by 7 and 9 from 1 to n

```python
n = int(input("Enter a number (n): "))

sum_divisible = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        sum_divisible += i

print(f"The sum of numbers divisible by both 7 and 9 from 1 to {n} is: {sum_divisible}")
```

### Assignment 5.3: Add all prime numbers from 1 to n

```python
n = int(input("Enter a number (n): "))

total_prime_sum = 0

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        total_prime_sum += num

print(f"The sum of all prime numbers from 1 to {n} is: {total_prime_sum}")
```

### Assignment 6.1: Add all odd numbers from 1 to n using a function

```python
def add_odd_numbers(n):
    total_sum = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total_sum += i
    return total_sum

n_odd = int(input("Enter a number "))
result_odd = add_odd_numbers(n_odd)
print(f"The sum of odd numbers: {result_odd}")
```

### Assignment 6.2: Add all prime numbers from 1 to n using a function

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def add_prime_numbers(n):
    total_prime_sum = 0
    for num in range(2, n + 1):
        if is_prime(num):
            total_prime_sum += num
    return total_prime_sum

n_prime = int(input("Enter a number "))
result_prime = add_prime_numbers(n_prime)
print(f"The sum of all prime : {result_prime}")
```
