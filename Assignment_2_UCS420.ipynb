```python
import math

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(math.sqrt(num)) + 1):
        if num % i == 0:
            return False
    return True
```

```python
# iv. Count and print all prime numbers, and build the list of primes
prime_numbers_list = [num for num in random_numbers if is_prime(num)]
prime_count = len(prime_numbers_list)
print(f"Count of prime numbers: {prime_count}")
print("List of prime numbers:", prime_numbers_list)
```

```python
# Using the same 8 digits of your roll number from Q1
# Ensure roll_number_str is available and has at least 8 digits
if len(roll_number_str) >= 8:
    roll_digits = [int(digit) for digit in roll_number_str[:8] if digit.isdigit()]
    A = {digit * 7 for digit in roll_digits}
    B = {digit * 9 for digit in roll_digits}
    print("Set A:", A)
    print("Set B:", B)
else:
    print("Roll number string is too short to create sets A and B with 8 digits.")
    # Fallback for demonstration if roll_number_str is shorter than 8 digits
    roll_digits = [int(digit) for digit in roll_number_str if digit.isdigit()]
    A = {digit * 7 for digit in roll_digits}
    B = {digit * 9 for digit in roll_digits}
    print("Set A (from available digits):", A)
    print("Set B (from available digits):", B)
```

```python
# vi. Find the union of A and B
union_AB = A.union(B)
print("Union of A and B (A | B):", union_AB)
```

```python
# vii. Find the intersection of A and B
intersection_AB = A.intersection(B)
print("Intersection of A and B (A & B):", intersection_AB)
```

```python
# viii. Find A - B and B - A separately using difference()
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)
print("A - B (A.difference(B)):", diff_A_B)
print("B - A (B.difference(A)):", diff_B_A)
# difference() shows elements unique to one set, symmetric_difference() shows elements unique to either set (union - intersection).
```

```python
# ix. Find the symmetric difference of A and B
symmetric_diff_AB = A.symmetric_difference(B)
print("Symmetric difference of A and B (A ^ B):", symmetric_diff_AB)
```

```python
# x. Check whether A is a subset of B, and whether B is a superset of A
is_A_subset_B = A.issubset(B)
is_B_superset_A = B.issuperset(A)
print(f"Is A a subset of B? {is_A_subset_B}")
print(f"Is B a superset of A? {is_B_superset_A}")
```

```python
# xi. Ask the user to input a value X. Remove X from set A if present using discard()
user_input_X_str = input("Enter a value to discard from Set A: ")

user_input_X = int(user_input_X_str)
A.discard(user_input_X) # discard() removes an element if present, does nothing if not present.
print(f"Set A after discarding {user_input_X}:", A)
```

```python
# 5. Consider the following dictionary:
my_dict = {
    "name": "Your Name",
    "roll_no": roll_number_str,
    "branch": "Your Branch",
    "age": 20,
    "city": "Your Home City"
}
print("Original dictionary:", my_dict)
```

```python
# i. Rename the key “city” to “location”

# Using pop() to rename 'city' to 'location'

city_value = my_dict.pop("city")
my_dict["location"] = city_value
print("Dictionary after renaming 'city' to 'location':", my_dict)
```

```python
# ii. Add a new key “cgpa”
my_dict["cgpa"] = 9.5
print("Dictionary after adding 'cgpa':", my_dict)
```

```python
# iii. Update the value of “age” by increasing it by 1

my_dict["age"] += 1
print("Dictionary after updating 'age':", my_dict)
```

```python
# iv. Delete the “branch” key using two different methods

# Method 1: Using pop()
dict_copy_pop = my_dict.copy()

popped_value = dict_copy_pop.pop("branch")
print(f"Dictionary after pop() on 'branch': {dict_copy_pop}, Popped value: {popped_value}")

# Method 2: Using del keyword
dict_copy_del = my_dict.copy()

del dict_copy_del["branch"]
print("Dictionary after del on 'branch':", dict_copy_del)

```

```python
# v. Iterate over your dictionary using .items() and print each key-value pair
print("Iterating through dictionary items:")
for key, value in my_dict.items():
    print(f"{key} -> {value}")
```

```python
# vi. Check for "email" key existence and print a safe fallback message
if "email" in my_dict:
    print(f"Email: {my_dict['email']}")
else:
    print("Email key not found. Using fallback message: 'No email provided'.")
```

```python
# vii. Create a second dictionary friend_dict and merge it
friend_dict = {
    "name": "Friend's Name",
    "roll_no": "F123456789",
    "branch": "Friend's Branch",
    "age": 21,
    "location": "Friend's City",
    "cgpa": 8.8
}
merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
# When both dictionaries share a key, the value from the second dictionary (friend_dict) "wins" or overwrites the value from the first dictionary (my_dict).
```

```python
# viii. Use dictionary comprehension to create a new dictionary with only string values
string_values_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print("New dictionary with only string values:", string_values_dict)
```
